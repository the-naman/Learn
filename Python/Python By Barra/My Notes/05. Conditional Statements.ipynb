# Conditional Statements : 

```
Checkpoint that checks a condition : 
    If True : Run the Code
    If False : Skip it
```

# Standalone If : If this is True, do this, otherwise, do nothing.

## Chart :

## Syntax : 

```
if (condition):
    next_statement
```

## Practice : 

````python
sun = True
if sun == True:
    print('Day')
if sun == False:
    print('Night')
````

# Else : Runs only if all previous conditions are False, "if nothing was True do this instead"

## Chart : 

## Practice : 

````python
# Day-Night problem
sun = False
if sun == True:
    print('Day')
else:
    print('Night')
````

### 1. Number Checking (+, -, 0) : 

````python
# Check number is Positive, Negative, Zero
a = 11
if a == 0:
    print('Number is zero.')
elif a < 0:
    print('Number is negative.')
else:
    print('Number is positive and greater than zero.')
````

### 2. Grading System : 

````python
score = 95

if score >= 90:
    print('A')
elif score >= 80:
    print('B')
elif score >= 70:
    print('C')
elif score >= 60:
    print('D')
elif score < 60 and score >= 0:
    print('F')
else:
    print('Enter valid score.')
````

# Nested If : if statement inside another if, if the first is True, then check the second

## Chart : 

## Practice : 

### 1. Grading with or without project.

````python
score = 79
project = True

if score >= 90:
    if project:
        print('A+')
    else:
        print('A')

elif score >= 80:
    if project:
        print('B+')
    else:
        print('B')

elif score >= 70:
    if project:
        print('C+')
    else:
        print('C')

elif score >= 60:
    print('D')

elif score < 60:
    print('F')

else:
    print('Enter a valid score.')
````
