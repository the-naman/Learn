# Conditional Statements : 

```
Checkpoint that checks a condition : 
    If True : Run the Code
    If False : Skip it
```

# Standalone If : If this is True, do this, otherwise, do nothing.

## Chart :

## Syntax : 

```
if (condition):
    next_statement
```

## Practice : 

````python
sun = True
if sun == True:
    print('Day')
if sun == False:
    print('Night')
````

# Else : Runs only if all previous conditions are False, "if nothing was True do this instead"

## Chart : 

## Practice : 

````python
# Day-Night problem
sun = False
if sun == True:
    print('Day')
else:
    print('Night')
````

````python
score = 95

if score >= 90:
    print('A')
elif score >= 80:
    print('B')
elif score >= 70:
    print('C')
elif score >= 60:
    print('D')
elif score < 60 and score >= 0:
    print('F')
else:
    print('Enter valid score.')
````

# Nested If : if statement inside another if, if the first is True, then check the second

## Chart : 

## Practice : 

### 1. Grading with or without project.

````python
score = 79
project = True

if score >= 90:
    if project:
        print('A+')
    else:
        print('A')

elif score >= 80:
    if project:
        print('B+')
    else:
        print('B')

elif score >= 70:
    if project:
        print('C+')
    else:
        print('C')

elif score >= 60:
    print('D')

elif score < 60:
    print('F')

else:
    print('Enter a valid score.')
````

# Inline If (Ternary) : 

## Chart : 

## Syntax : 

```
A if condition else B
```

## Practice : 

````python
# Even Odd using Ternary or Inline IF-ELSE
x = 22
print(x, ': is an Even Number') if x % 2 == 0 else print(
    x, ': is an Odd Number')
````

````python
# Positive Negative Using Inline IF ELSE
y = 1
print(y, ': is Negative Number') if y < 0 else print(
    y, ': is Positive Number')
````

````python
# Positive Negative Using Inline IF ELSE in best formatting
a = 2
print(f'{a}, is Negative Number'
      if a < 0 else
      f'{a}, is Positive Number')
````

````python
# Positive Negative Using Inline IF ELSE using variable
b = 10
pn = "Negative Number" if b < 0 else "Positive Number"
print(f'{b} : is a {pn}')
````

````python
# Positive Negative and Zero Using Inline IF ELSE using Nesting IF ELSE
x = 10
pnz = 'Negative Number' if x < 0 else 'Zero' if x == 0 else 'Positive Number'
print(f'{x} : {pnz}')
````

# Match Case : Evaluate a value against multiple values Runs the code of the first match
```
Only works for matching cases
Don't work for greater or lower case
```

## Practice

### Convert the full country names into 2-letter abbreviations

````python
# Using IF ELIF ELSE
country = 'INDIA'
if country == 'United States':
    print('US')
elif country == 'INDIA':
    print('IN')
else:
    print('Unknow Value')
````

````python
# Using Match Case :
country = input('Enter the country name: ')
match country:
    case 'United States' | 'USA':
        print('US')
    case 'INDIA' | 'BHARAT':
        print('IN')
    case 'Canada':
        print('CA')
    case 'Switezerland':
        print('SW')
    case 'United Kingdom':
        print('UK')
    case 'France':
        print('FR')
    case 'Ireland':
        print('IR')
    case _:
        print('Unkown Value')
````

## 1. Validate the quality and correctness of email values.
```
Conditions : 
    1. Must not be empty.
    2. Must contain '.' and '@'.
    3. Must contain exactly only one '@' symbol.
    4. Must ends with '.com', '.org' or '.net'.
    5. Must not be longer than 254 characters.
    6. Must start and ends with a letter or digit.
```

````python
# using independent IF :

# email
email = 'bh@gmail.net'
# ----------------------------------------------------------------------------------------------
# Email string clean-up
email = email.strip()
# ----------------------------------------------------------------------------------------------
# Email never be empty
if email == '':
    print('Empty email not allowed.')
# ----------------------------------------------------------------------------------------------
# Email always contains '.' and '@'
if not ('.' in email and '@' in email):
    print('Email must contains . and @.')
# ----------------------------------------------------------------------------------------------
# Email doesn't contains '@' more than 1
if email.count('@') != 1:
    print("Email must conatins only one @.")
# ----------------------------------------------------------------------------------------------
# Email always have a domain extension like .com, .org, .net
if not (email.endswith('.com') or email.endswith('.net') or email.endswith('.org')):
    print('Email Domain Extension are not allowed')
# ----------------------------------------------------------------------------------------------
# Email length never should be greater than 254 charcaters
if len(email) > 254:
    print('Invalid Length of email')
# ----------------------------------------------------------------------------------------------
# Email always start and end with letters or digits
if not (email[0].isalnum() and email[-1].isalnum()):
    print('Email only starts and ends with alphabets or digits.')
# ----------------------------------------------------------------------------------------------
# Email output
print(f'Email : {email}')
````

````python
# using print :

email = 'abc1@gmail.com'
# -----------------------------------------------------------------------------------------------
# email attributes :
email_length = len(email)
dot_count = email.count('.')
At_count = email.count('@')
domain_extension = email[email.rindex('.'):]
# -----------------------------------------------------------------------------------------------
# solutions
one = len(email) > 0
two = '.' in email and '@' in email
three = email.count('@') == 1
four = email.endswith('.com') or email.endswith(
    '.org') or email.endswith('.net')
five = len(email) <= 254
six = email[0].isalnum() and email[-1].isalnum()
# -----------------------------------------------------------------------------------------------
# solution output
print('Conditions need to be TRUE always: ')
print(f'\tEmail should not be empty or blank              : {one}')
print(f'\tEmail length should always greater than 0       : {two}')
print(f'\tEmail contains only one @                       : {three}')
print(f'\tEmail ends with only .com, .org, .net           : {four}')
print(f'\tEmail length not be longer that 254 characyers  : {five}')
print(f'\tEmail only starts and ends with letters or digits : {six}')
# -----------------------------------------------------------------------------------------------
# attributes output
print('\nEmail Details              : ')
print(f'\tEmail Length              : {email_length} charcater')
print(f"\t'.' count                 : {dot_count} character")
print(f"\t'@' count                 : {At_count} character")
print(f"\tEmail Domain Extension    : {domain_extension}")
````
