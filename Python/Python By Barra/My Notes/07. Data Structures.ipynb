##### INDEXING : Accessing and Reading Matrix List

````python
matrix = [['a', 'b', 'c'],
          [1, 2, 3],
          [True, None, 3.5, 1+9j, (), {}]]

# accessing and reading each element of matrix

# printing Matrix aand its type
print('Level 1 : Printing Whole Matrix : ')
print(f'Matrix : {matrix}')
print(f'Type of Matrix : {type(matrix)}')
print(f"{'-'*90}\n")

# printing each list and their type under whole matrix
print('Level 2 : All lists under Matrix : ')
for l in matrix:
    print(f'\tType : {type(l)}    and    List = {l}')
print(f"{'-'*90}\n")

# accessing and printing elements from each lists and the type of them
print('Level 3 : Printing element from list of Matrix : ')
for l in matrix:
    print(f'\nList : {l}')
    for item in l:
        print(
            f'\titem = {item}       and          Type : {type(item)}        ')
print(f"{'-'*90}\n")
````

##### SLICING : Slicing from matrix

````python
matrix = [[1, 2, 3, 4, 5],
          ['a', 'b', 'c', 'd', 'e'],
          [True, None, 3.5, 1+97j, (), {}]]

print(f'Matrix = {matrix}')
print(f'{"-"*100}\n')

print(f'1. Sliced out first two list from matrix')
print(matrix[:2])
print(f'{"-"*100}\n')

print(f'2. Sliced out last two list from matrix')
print(matrix[1:])
print(f'{"-"*100}\n')

print(f'3. Sliced out middle list from matrix')
print(matrix[1:2])
print(f'{"-"*100}\n')
````

##### SLICING : Slicing from nested list of matrix

````python
matrix = [[1, 2, 3, 4, 5],
          ['a', 'b', 'c', 'd', 'e'],
          [True, None, 3.5, 1+97j, (), {}]]

print(f'Matrix = {matrix}')
print(f'{"-"*100}\n')

print(f'1. Sliced out first two element from first list of matrix')
print(matrix[0][:2])
print(f'{"-"*100}\n')

print(f'2. Sliced out last two element from last list of matrix')
print(matrix[-1][-2:])
print(f'{"-"*100}\n')

print(f'3. Sliced out all element except fist and last element from middle list of matrix')
print(matrix[1][1:-1])
print(f'{"-"*100}\n')
````

## Unpacking : 

````python
person = ['Maria', 29, 'Data Engineer', 'Spain']
print(f'List -----> \n\tperson = {person}')
print('-' * 70, '\n')

# Method : 01 : Traditional Method
print('Method 01 : Using Indexes...........')
print(f'\tname      = {person[0]}')
print(f'\tage       = {person[1]}')
print(f'\trole      = {person[2]}')
print(f'\tcountry   = {person[3]}')
print('-' * 70, '\n')

# Method : 02 : Easy Way
print('Method 02 : Using Unpacking Method..')
name, age, role, country = person
print(f'\t name, age, role, country = {name}, {age}, {role}, {country}')
print('-' * 70, '\n')
````

### REST COLLECTOR * or Star Operator : 

#### Chart : 

#### Practice : 

````python
person = ['Deo', 24, 'Army', 'India']
print(f'List ---> \n\tperson = {person}')
print(f"{'-' * 71}\n")

# I have to extract only Name and Country
# Unpacking using * Method
print(f'Unpacking Using REST Method...')
name, *details, country = person
print(f'name, details, country = {name}, {details}, {country}')
````

````python
person = ['Anjali', 23, 'Model', 'Mumbai', 'India']
name, *details = person
print(name)
print(details)
````

````python
person = ['Mihir', 27, 'Pilot', 'Pune', 'India']
*details, country = person
print(country)
print(details)
````

### Unpacking Rules : 

```
1. Number of VARIABLES = Number of VALUE : MUST
2. If in list currently one value then you may use as ONE VARIABLE and *VARIABLE : NO ANY ERROR
3. If you pass string : 'Hi' : in option 2 it will store 'H' in 1st variable and 'i' in * variable as 'H' and ['i']
```

````python
number1 = [1, 5, 8, 2, 3]
number2 = [1, 0, 1, 2, 5]

str1 = ['a', 'b', 'c']
str2 = ['a', '', 'b']
str3 = ['a', ' ', 'c']
str4 = ['a', None, 'm']

print(
    f'All ists are as below:\n\tnumber1 = {number1}\n\tnumber2 = {number2}\n\n\tstr1 = {str1}\n\tstr2 = {str2}\n\tstr3 = {str3}\n')

print(f'All (number1) : {all(number1)}')
print(f'All (number2) : {all(number2)}   # Holding 0\n')
print(f'All (str1) : {all(str1)}')
print(f'All (str2) : {all(str2)}      # Holding empty string')
print(f'All (str3) : {all(str3)}       # Holding blank space string')
print(f'All (str4) : {all(str4)}      # Holding None')
````

### Comparision Operator : 

````python
l1 = [1, 2, 3]
l2 = [1, 2, 3]
l3 = [3, 1, 2]
l4 = [5, 8, 9, 3]
l5 = [5, 8, 0, 7]
l6 = [5, 8, 9, 9]

all_list = (l1, l2, l3, l4, l5, l6)

print('All list are:')
for index, lists in enumerate(all_list, start=1):
    print(f'\tl{index} = {lists}')

print(f"{'-' * 71}\n")

print(f'l1 == l2 : {l1 == l2}')
print(f'l2 < l3  : {l2 < l3}')
print(f'l4 < l5  : {l4 < l5}')
print(f'l5 > l6  : {l5 > l6}')
print(f'l1 != l2 : {l1 != l2}')
````

#### .insert(place, value) : insert value on the given position in the list.

````python
str1 = ['m', 'n', 'p', 'q']
print(f'str1 = {str1}\n')

print(f"Adding : \n\t'A' at the starting position\n\t'O' in between of n and p\n\t'Z' at the last position")

str1.insert(0, 'A')
print(f'str1 = {str1}\n')

str1.insert(str1.index('n')+1, 'O')
print(f'str1 = {str1}\n')

str1.insert(-1, 'Z')
print(f'str1 = {str1} : So here we can see that -1 index is not working\n')

str1.insert(len(str1)+1, 'Z')
print(f'str1 = {str1} : Here using len()+1 formula')
````

````python
import math

sepa = '-' * 150

matrix = [[1, 2, 3],
          ['a', 'b', 'c', 'd'],
          [None, 1+8j, (), {}, True]]
print(f'matrix = {matrix}\n{sepa}')

print(f"""\nTask : 
    1. Add a list [(1, 2), {'a', 'b'}, [None, False, {{}}, ()]] in middle of matrix
    2. Add 'False' after 'True' in matrix last list
    3. Convert 'c' to 'X' in the given matrix list
    4. Add '0' before '1' in the given matrix
    5. Find out type of second item in last list of matrix
    6. Add '4' at last of first list in the given matrix
{sepa}

Solutions : 
{sepa}
""")


# Solutions
# 1 : Adding List : [(1, 2), {'a', 'b'}, [None, False, {}, ()]] at Middle of matrix
one = matrix.insert((math.ceil(len(matrix)/2)),
                    [(1, 2), {'a', 'b'}, [None, False, {}, ()]])
print(f'1. List Added : {matrix}\n')

# 2 : Adding False at last of last list of matrix
two = matrix[-1].insert(len(matrix[-1])+1, False)
print(f'2."False" Added : {matrix}\n')

# 3 : Replcing first occurance of 'c' to 'X'
three = matrix[1][2] = 'X'
print(f"3.'c' replaced with 'X' : {matrix}\n")

# 4
four = matrix[0].insert(0, 0)
print(f"4.'0' instered at first : {matrix}\n")

# 5
five = type(matrix[-1][1])
print(
    f"5.Type of second item - {matrix[-1][1]}, in last list of matrix is : {five}\n")

# 6
six = matrix[0].append(4)
print(f"6.'4' inserted : {matrix}")
````
