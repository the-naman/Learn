# Data Structure : 

## What is Data Structure?

```
Data Structure is a way of organising and storing data so it can be used effciently

It python there are 4 Built-in Data Structures:
    1. list[]
    2. tuple()
    3. set{}
    4. dict{}
```

## Why do we need Data Structure?

```
Helps in :
    1. Organise    : in best format or way
    2. Manage      : stored and manage large scale of data with easy
    3. Manipulate  : do many operations on those data easily using built-in functions
```

### [] : 

````python
empty = []
print(f'empty = {empty}')
print(f'Type of list = {type(empty)}')
print('-' * 40, '\n')

letters = ['a', 'b', 'c', 'd', 'e']
print(f'letters = {letters}')
print(f'Type of list = {type(letters)}')
print('-' * 40, '\n')

numbers = [1, 2, 3, 4, 5]
print(f'numbers = {numbers}')
print(f'Type of list = {type(numbers)}')
print('-' * 40, '\n')
````

### list() :

````python
empty = list()
print(f'empty = {empty}')
print(f'Type of empty = {type(empty)}')
````

````python
letters = list('Python')
print(f'letters = {letters}')
print(f'Type of letters = {type(letters)}')

print(f'\nLetter from Letters : {letters[0]}')
print(f'Type of letters = {type(letters[0])}')
````

````python
numbers = list(range(1, 11))
print(f'numbers = {numbers}')
print(f'Type of numbers = {type(numbers)}')

print(f'\nNumber in numbers = {numbers[0]}')
print(f'Type of Number in numbers = {type(numbers[0])}')
````
