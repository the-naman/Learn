````python
# normal comparision :
print(10 == 10)
print(10 != 10)
print(7 > 3)
print(7 >= 7)
print(4 < 9)
print(5 <= 5)
````

````python
# chain comparision :
# it is from left to right one by one
print(1 < 4 < 6)
print(1 > 3 < 4)

print('a' == 'a' > 2)
````

## Identity Operators : 

### is : checks if two variables refer to the same object in memory

#### Chart : 

#### Practice : 

````python
x = ['a', 'b', 'c']
y = ['a', 'b', 'c']

print(f'x ID = {id(x)}')
print(f'y ID = {id(y)}')

print(x == y)
print(x is y)
````

````python
a = 101
b = 101

print(f'a ID = {id(a)}')
print(f'b ID = {id(b)}')

print(a == b)
print(a is b)
````

````python
x = ['a', 'b', 'c']
y = x

print(f'x ID = {id(x)}')
print(f'y ID = {id(y)}')

print(x == y)
print(x is y)
````
