### == , !=, <, >, <=, >= : 

````python
# normal comparision :
print(10 == 10)
print(10 != 10)
print(7 > 3)
print(7 >= 7)
print(4 < 9)
print(5 <= 5)
````

````python
# string or characyer comparision :
print('a' == 'a')
print('a' < 'b')
print('a' == 'A')
print('b' < 'A')
````

````python
# chain comparision :
# it is from left to right one by one
print(1 < 4 < 6)
print(1 > 3 < 4)

print('a' == 'a' > 2)
````

## Logical Operators : 

### and : used to combine multiple boolean expression, all condition should be True

````python
print(3 < 5 and 5 == 5)
````

````python
print(3 > 5 and 5 == 5)
````

````python
# id password or login
email = True
password = False
print(email and password)
````

````python
# id password or login
email = True
password = True
print(email and password)
````

### or : used to combine multiple boolean expression, atleast one condition must be True

````python
print(3 < 5 or 5 == 5)
````

````python
print(3 > 5 or 5 == 5)
````

````python
print(3 > 5 or 5 != 5)
````

````python
# consulmption meter monitoring
cpu_usage = 54
memory_usage = 95
print(cpu_usage > 90 or memory_usage > 90)
````

````python
# consumption meter monitoring
cpu_usage = 54
memory_usage = 65
print(cpu_usage > 90 or memory_usage > 90)
````
