#### Printing multiple rounds.

````python
print('Round : 1')
print('Round : 2')
print('Round : 3')
print('Round : 4')
print('Round : 5')
````

````python
# for using set
items = (1, 2, 3, 4, 5)
for item in items:
    print(f'Round : {item}')
````

````python
# for using list
items = [1, 2, 3, 4, 5, 'End']
for item in items:
    print(f'Round : {item}')
````

````python
# for using string
for char in 'Strings':
    print(f'Char : {char}')
````

````python
# for using range(stop)
items = range(5)
for item in items:
    print(item)
````

````python
# for using range(start, stop)
items = range(1, 5)
for item in items:
    print(item)
````

````python
# for using range(start, stop, step) -----> EVEN NUMBERS
items = range(2, 10, 2)
for item in items:
    print(item)
````

````python
# for using range(start, stop, step) -----> ODD NUMBERS
items = range(1, 10, 2)
for item in items:
    print(item)
````

##### We use FOR loops to go through values and aggregate data like summing, counting, or averaging.

````python
scores = [80, 50, 60, 75]
total = 0
count = 0
avg = 0

for score in scores:
    total += score
    count += 1
    avg = (avg + score)/count
    print(f'Current Total = {total}')
    print(f'Current Count = {count}')
    print(f'Current Avg = {avg}')
    print('-' * 40)
    print()
````

##### We use FOR loops to transform data like cleaning data before processing.

````python
files = [' Report.csv', 'DATA.csv ', ' final.TXT']
cleaned_files = []
print(f'Given List : {files}\n')
print('Final Files : ')
for file in files:
    file = file.strip().lower().replace('txt', 'csv')
    print('\t', file)
    cleaned_files.append(file)
print(f'\nUpdated List : {cleaned_files}')
````

## Control Statements : 

### Break : 
```
It stops the loop immediately it jumps out and ends the loop right away.
```

#### Chart : 

#### Practice : 

````python
names = ['John', 'Mohini', '', 'Kumar']
for name in names:
    if name == '':
        print('Empty value detected!!')
        break
    print(f'Name = {name}')
````

### Continue : 
```
It skips one loop cycle without stopping the loop.
Means skip this one and go to next
```

#### Chart : 

#### Practice : 

````python
items = ['Sonu', 'Anjali', '', 'Radha']
for item in items:
    if item == '':
        print(f'!! Empty Name Skipped !! at INDEX : {items.index(item)}')
        continue
    print(item)
````

### Pass : 
```
It is a placeholder where nothing happens.
For now... Just keeps it and Do nothing
```
