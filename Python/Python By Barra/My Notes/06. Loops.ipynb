##### We use FOR loops to go through values and aggregate data like summing, counting, or averaging.

````python
scores = [80, 50, 60, 75]
total = 0
count = 0
avg = 0

for score in scores:
    total += score
    count += 1
    avg = (avg + score)/count
    print(f'Current Total = {total}')
    print(f'Current Count = {count}')
    print(f'Current Avg = {avg}')
    print('-' * 40)
    print()
````

##### We use FOR loops to transform data like cleaning data before processing.

````python
files = [' Report.csv', 'DATA.csv ', ' final.TXT']
cleaned_files = []
print(f'Given List : {files}\n')
print('Final Files : ')
for file in files:
    file = file.strip().lower().replace('txt', 'csv')
    print('\t', file)
    cleaned_files.append(file)
print(f'\nUpdated List : {cleaned_files}')
````

## Control Statements : 

### Break : 
```
It stops the loop immediately it jumps out and ends the loop right away.
```

#### Chart : 

#### Practice : 

````python
names = ['John', 'Mohini', '', 'Kumar']
for name in names:
    if name == '':
        print('Empty value detected!!')
        break
    print(f'Name = {name}')
````

### Continue : 
```
It skips one loop cycle without stopping the loop.
Means skip this one and go to next
```

#### Chart : 

#### Practice : 

````python
items = ['Sonu', 'Anjali', '', 'Radha']
for item in items:
    if item == '':
        print(f'!! Empty Name Skipped !! at INDEX : {items.index(item)}')
        continue
    print(item)
````

### Pass : 
```
It is a placeholder where nothing happens.
For now... Just keeps it and Do nothing
```

#### Chart : 

#### Practice : 

````python
items = ['Ram', 'Shyam', '', 'Anjali']
for item in items:
    if item == '':
        pass
    print(item)
````

### Break VS Continue : Real World Applications : 

#### Loop through a list of days and print only the working days, skipping the weekends.

````python
# CONTINUE EXAMPLE : Looping WORKDAYS ignoring WEEKNED
days = ['Sunday', 'Monday', 'Tuesday',
        'Wednesday', 'Thrusday', 'Friday', 'Saturday']
weekends = ['Sunday', 'Saturday']
for day in days:
    if day in weekends:
        continue
    print(day)
````

#### Scan emails to block unsafe data from entering your systems

````python
# BREAK EXAMPLE : Scanning safe email and stopping unsafe email from accessing data
# Using constants
BOLD = "\033[1;3m"
RESET = "\033[0m"

emails = ['naman@gmail.com', 'python@gmail.com',
          'baraa@yahoomail.com', 'realmail@realemail.com', 'sql@outlook.com', 'google.com', 'DROP TABLE FROM USERS;', 'data@hotmail.com']
safe_domain_extensions = ['gmail.com',
                          'yahoomail.com', 'outlook.com', 'hotmail.com']
sep = '@'
for email in emails:
    # this protects from SQL injection or bad email
    if ';' in email:
        print(
            f'\n{BOLD}CRITICAL : SQL INJECTION ATTACK HAPPENED : {email}{RESET}\n')
        break
    elif sep not in email:
        print(f'\nUNSAFE  : Invalid Email                 : {email}')
    else:
        # this protects from bad domain
        domain = email.split('@', 1)[1]
        if domain in safe_domain_extensions:
            print(f'SAFE    : Login Successful with         : {email}')
        else:
            print(f'\nUNSAFE  : Bad or Invalid Email Domain   : {domain}\n')
````

### For - Else : 
```
Runs a block of code only if the loop finishes naturally
Loops completed without breaks
```

#### Chart : 

#### Practice : 

````python
numbers = [1, 2, 3, 4]
for number in numbers:
    print(number)
else:
    print("Number Printing Ended.")
````

### For - Else + Break : 

#### Chart : 

#### Practice : 

````python
numbers = [1, 2, 3, 4, 5, 6, 7]
for number in numbers:
    if number == 5:
        break
    print(number)
else:
    print('Successfully this execution round completed.')
````

Check if all files are CSV files if not then convert that first.

````python
files = ['report.csv', 'sales.csv', 'supply.csv', 'admin.txt', 'finance.csv']
non_authorize_file_format = ['txt', 'pdf']

for file in files:
    if file.split('.', 1)[1] in non_authorize_file_format:
        print(
            f'{file.replace(
                file.split('.', 1)[0], len(file.split('.', 1)[0]) * '*')} is not a CSV file, first fix this file before moving ahead.')
        break
    print(file)
else:
    print('All files are of CSV format.')
````

##### 1. Crossing Data Example

##### 2. Creating Report for per day in CSV : Heirarichal Data

````python
years = [2026]
months = ['Jan', 'Feb', 'Mar']
days = range(1, 7)

for year in years:
    for month in months:
        print(f"{month}:{year} - Reports\n{'*'*30}")
        for day in days:
            print(f'Report_{day}_{month}_{year}.csv')
        print()
````

##### 3. create query for SQL for multiple tables and count the customer where have NULL id.

````python
# FORMAT : SELECT COUNT(*) FROM table_Name WHERE id IS NULL;
tables = ['customers', 'orders', 'products', 'prices']
columns = ['id', 'create_date']

for table in tables:
    print(f"Table : {table.upper()}\n{'*'*30}")
    for column in columns:
        print(f'SELECT COUNT(*) FROM {table} WHERE {column} IS NULL;')
    print(f'{"="*60}\n')
````
