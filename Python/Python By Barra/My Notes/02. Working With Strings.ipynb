# Data Transformation :

## f{} - string: it is a modern, super-easy way to format and build strings, with this we can put variables, expressions directoly into the string

````python
name = 'Sam'
age = 34
is_student = False


# using '+' operator and this is hard method
print("My name is " + name + ", I am " + str(age) +
      " years old and my student status is " + str(is_student) + '.')


# using f-string and this is easy method
print(
    f"My name is {name}, I am {age} years old and my student status is {is_student}.")
````

## replace() : swaps part of the text with something new

### Syntax :

```
text.replace(old, new)
```

### Practice : 

````python
price = '12345,98'
print('Old Format of Price:', price)
print('New Format of Price:', price.replace(',', '.'))
print("',' replcaed with '.'")
````

````python
# Date
d = '2026/08/15'
print(d)
# replace / with -
print(d.replace('/', '-'))
````

````python
# Mobile Number
MN = '987-853-1764'
print(MN)
# remove the - from whole mobile number
print(MN.replace('-', ""))
````

````python
price = '$1,265.99'
print(price)

# we want price as float without any sign or comma
print(price.replace('$', '').replace(',', ''))
````

### Challenge :

#### Convert this "+49 (176) 123-4567" to "00491761234567".

````python
pn = "+49 (176) 123-4567"
print('Phone Number =', pn)

print('Number:', pn.replace(
    '+', '00').replace(' (', '').replace(') ', '').replace('-', ''))
````

## split() : breaks a string into smaller parts

````python
stamp = '2026-08-12 15:45:23'
print(f'stamp: {stamp}')

dt = stamp.split(' ')
print(f'Date and Time: {dt}')

d = dt[0]
print(
    f"Date: {d} ; ---->  Year: {d.split('-')[0]} ; Month: {d.split('-')[1]} ;   Date: {d.split('-')[2]}")

t = dt[1]
print(
    f'Time: {t} ;   ---->  Hours: {t.split(':')[0]} ;  Minutes: {t.split(':')[1]} ; Seconds: {t.split(':')[2]}')
````

````python
csv_record = "1234,Max,USA,1970-10-05,M"
print(
    f"ID: {csv_record.split(',')[0]}\nName: {csv_record.split(',')[1]}\nCountry: {csv_record.split(',')[2]}\nDoB: {csv_record.split(',')[2]}\nGender: {csv_record.split(',')[3]}")
````
