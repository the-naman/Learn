### Print the following using variable.
```
My name is Naman.
Naman is learning Python.
Naman wants to become Python expert.
```

````python
name = 'Naman'
print('My name is', name)
print(name, 'is learning Python.')
print(name, 'wants to become Python expert.')
````

### Now if some one wants to change name i have to just change once instead of multiple times. Change it to name as KN.

### Word python also occuring multiple times and it is subject or language so may be in future i have to replace. So create a subject variable for it.

````python
name = 'Naman'
subject = 'Python'
print('My name is', name, '\b.')
print(name, 'is learning', subject, '\b.')
print(name, 'wants to become', subject, 'expert.')
````

##### Function Example.

````python
text = 'Hi'
number = 10

print('Printing Text:', text)
print('Printing Number:', number)

print('Printing TYPE of text:', type(text))
print('Printing TYPE of number:', type(number))

print('Printing length of text', len(text))
# print('Printing length of number', len(number))            # len() cannot work on int data type
print('Printing length of Number:', len(str(number)))
````

````python
text = 'Hi'
number = 10

print(text.upper())
# print(number.upper())             .upper() method cannot work for int or number

# print(text.bit_length())          .bit_length() method cannot works for str
print(number.bit_length())
````
