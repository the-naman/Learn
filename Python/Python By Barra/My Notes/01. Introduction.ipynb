## Challenges :

### Print all of following without variable.
```
My name is Naman.
Naman is learning Python.
Naman wants to become Python expert.
```

````python
print('My name is Naman.')
print('Naman is learning Python.')
print('Naman wants to become Python expert.')
````

### Print the following using variable.
```
My name is Naman.
Naman is learning Python.
Naman wants to become Python expert.
```

````python
name = 'Naman'
print('My name is', name)
print(name, 'is learning Python.')
print(name, 'wants to become Python expert.')
````

### Now if some one wants to change name i have to just change once instead of multiple times. Change it to name as KN.

````python
name = 'KN'
print('My name is', name)
print(name, 'is learning Python.')
print(name, 'wants to become Python expert.')
````

### Word python also occuring multiple times and it is subject or language so may be in future i have to replace. So create a subject variable for it.

````python
name = 'Naman'
subject = 'Python'
print('My name is', name, '\b.')
print(name, 'is learning', subject, '\b.')
print(name, 'wants to become', subject, 'expert.')
````

### Print the following :
```
info@datawithbaraa.com
support@datawithbaraa.com
www.datawithbaraa.com
```

````python
domain = 'datawithbaraa.com'
print('info@', domain, sep='')
print('support@', domain, sep='')
print('www.', domain, sep='')
````

````python
domain = 'datawithbaraa.com'
print(f'info@{domain}')
print(f'support@{domain}')
print(f'www.{domain}')
````
