### When to use what?

### *args : it accepst multiple single data type of values

````python
# calculate the total of values
def total(a=0, b=0, c=0, d=0):
    print(a+b+c+d)


total(1, 2)
total(1, 2, 3)
total(1, 2, 3, 4)
````

````python
# calculate the total of values
def total(*args):
    print(type(args))


total()
````

````python
# calculate the total of values
def total(*args):
    print(f"Sum of {args[0]} to {args[-1]} = {sum(args)}")


total(1, 2)
total(1, 2, 3)
total(1, 2, 3, 4)
total(1, 2, 3, 4, 5, 6)
````

### **kwargs : it accepts multiple value or multiple data types

````python
# create user
def create_user(**kwargs):
    print(type(kwargs))


create_user()
````

````python
# create user
def create_user(**kwargs):
    print(f"User Created  : {kwargs}")


create_user(firstname='Som', lastname='Dutt', age=28, country='INDIA')
create_user(f_name='Madhuri', l_name='Dixit', country='IN')
create_user(name='Sonu Sood', age=44)
````

### Example 02:

````python
# Name clean up
def nameclean(name):
    name = name.strip()
    if not name:
        return 'Invalid Input'
    else:
        low_name = name.lower()
        up_name = name.upper()
        return low_name, up_name


clean_name = nameclean('  pUsHpa   ')
print(clean_name)
````

````python
# Given List
names = ['   miKA  ', '   AMAal', ' baBY    ', '     SONal', '   MohInI    ']
print(f"Original Name : {names}")

# creating list for new cleaned names
low_name = []
up_name = []

# loop for name clean up
for name in names:
    # using above user-defined function nameclean()
    low_Name, up_Name = nameclean(name)
    low_name.append(low_Name)
    up_name.append(up_Name)

# printing cleanup names
print(f"\nLower Case Names : {low_name}")
print(f"UPPER Case Names : {up_name}")
````
