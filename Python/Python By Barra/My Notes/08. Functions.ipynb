````python
# Traditional Name string clean up
print('Traditional Name string clean up......')
name = ' moHini '
print(f'name = {name}')
print('cleaned name =', name.strip().lower())
````

````python
# Name cleanup using function
print('Name cleanup using function......')


def name_cleanup(name):
    print(f"\tOriginal input string   : {name}")
    print(f"\tUsing capitalize()      : {name.strip().capitalize()}")
    print(f"\tUsing title()           : {name.strip().title()}")


name_cleanup('    kUmar naMAn')
````

### Local Variables : 

```
Created inside the function and can be accessed only indside the function
```

### Global Variables : 

```
Created outside of the function and can be accessed from anywhere throughout the program
```

### Practice : 

````python
greet_msg = 'Hello!!'                   # Global Variable


def greetwithname(name):                # Parameter : it also act as local variable
    cleaned_name = name.strip().title()  # Local Variable
    ask = 'How are you?'                # Local Variable
    print(f"{greet_msg} {cleaned_name}, {ask}")


greetwithname('kUMar naMAn         ')
````
