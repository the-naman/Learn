# Types : 

## type() : checks the data type of value

````python
x = 15
y = 3.14
z = 3 + 6j

print(f'x = {x} {type(x)}\
    \ny = {y} {type(y)}\
    \nz = {z} {type(z)}')
````

## int() : converts compatible value into integer value

````python
x = '24'
print(f'x = {x} {type(x)}')
print(f'x * 3 = {x*3}')
print('Changing type of x....')
x = int(x)
print(f'x = {x} {type(x)}')
print(f'x * 3 = {x*3}')
````

````python
y = 3.14
print(f'y = {y} {type(y)}')
print('Changing float to int...')

y = int(y)
print(f'y = {y} {type(y)}')
````

````python
z = 4 + 6j
print(f'z = {z} {type(z)}')
print('Changing to int....')

z = int(z.real)
print(f'z.real = {z} {type(z)}')
````

## float() : converts compatible value into float value

````python
a = '5.78'
print(f'a = {a} {type(a)}')
print('Converting str to float....')

a = float(a)
print(f'a = {a} {type(a)}')
````

````python
a = '5'
print(f'a = {a} {type(a)}')
print('Converting str to float....')

a = float(a)
print(f'a = {a} {type(a)}')
````

````python
b = 5
print(f'b = {b} {type(b)}')
print('Converting int to float....')

b = float(b)
print(f'b = {b} {type(b)}')
````

````python
c = 9 + 0j
print(f'c = {c} {type(c)}')
print('Converting complex number to float....')

c = float(c.real)
print(f'c = {c} {type(c)}')
````

## complex() : 

````python
e = '1'
print(f'e = {e} {type(e)}')
print('Converting str to complex....')

e = complex(e)
print(f'e = {e} {type(e)}')
````

````python
f = 3
print(f'f = {f} {type(f)}')
print('Converting int to complex....')

f = complex(f)
print(f'f = {f} {type(f)}')
````

````python
g = 7.98
print(f'g = {g} {type(g)}')
print('Converting float to complex....')

g = complex(g)
print(f'g = {g} {type(g)}')
````

````python
a = 5
b = 3
print(f'Complex of a = 5 and b = 3 is : {complex(a, b)}')
````

# Math : 

## Operators : + - * / // % **

````python
print(f'Addition        : 4+5       = {4+5}')
print(f'Subtraction     : 66-34     = {66-34}')
print(f'Multiplication  : 3*4       = {3*4}')
print(f'Division        : 44/11     = {44/11}')
print(f'Floor Division  : 44//11    = {44//11}')  # gives only int
print(f'Modulo          : 7%2       = {7 % 2}')
print(f'Exponent        : 2**3      = {2**3}')
````

## Round Chart :

## round() : rounds number to the nearest whole number up or down depending on what's closer

````python
price = 35.548798456
print(f'Input price = {price}\
    \nRounded price = {round(price)}')
````

````python
price = 35.548798456
print(f'Input price = {price}\
    \nRounded price double digit after decimal = {round(price, 2)}')
````

````python
price = 35.548798456
print(f'Input price = {price}\
    \nRounded price single digit after decimal = {round(price, 1)}')
````

```
Banker's Rounding : Ties (like 0.5) go to the nearest even number
```

````python
a = 1.5
b = 2.5
c = 3.5
d = 4.5
e = 5.5
f = 6.5
print(f'Values are :\
    \nGiven Input Value       : a = {a}   b = {b}     c = {c}     d = {d}     e = {e}     f = {f}\
    \
    \nBanker\'s Rounded Values : a = {round(a)}     b = {round(b)}\
       c = {round(c)}       d = {round(d)}       e = {round(e)}       f = {round(f)}')
````

## math.ceil() : rounds number up to the ceiling

````python
import math
a = 6.3
b = 6.5
c = 6.7
print(f'a = {a}\nb = {b}\nc = {c}\n\na ceil = {math.ceil(a)}\nb ceil = {math.ceil(b)}\nc ceil = {math.ceil(c)}')
````

## math.floor() : rounds number down to the floor

````python
import math
a = 4.3
b = 4.5
c = 4.8
print(f'a = {a}\nb = {b}\nc = {c}\n\na floor = {math.floor(a)}\nb floor = {math.floor(b)}\nc floor = {math.floor(c)}')
````

## math.trunc() : cut off the decimal part and keeps the whole number (no rounding)

````python
import math
price = 35.5567832
print(f'Price = {price}')
print(f'Truncated Price = {math.trunc(price)}')
print(f'Same but int Price = {int(price)}')
````
