# Types : 

## type() : checks the data type of value

```python
x = 15
y = 3.14
z = 3 + 6j

print(f'x = {x} {type(x)}\
    \ny = {y} {type(y)}\
    \nz = {z} {type(z)}')
```

## int() : converts compatible value into integer value

```python
x = '24'
print(f'x = {x} {type(x)}')
print(f'x * 3 = {x*3}')
print('Changing type of x....')
x = int(x)
print(f'x = {x} {type(x)}')
print(f'x * 3 = {x*3}')
```

```python
y = 3.14
print(f'y = {y} {type(y)}')
print('Changing float to int...')

y = int(y)
print(f'y = {y} {type(y)}')
```

```python
z = 4 + 6j
print(f'z = {z} {type(z)}')
print('Changing to int....')

z = int(z.real)
print(f'z.real = {z} {type(z)}')
```

## float() : converts compatible value into float value

```python
a = '5.78'
print(f'a = {a} {type(a)}')
print('Converting str to float....')

a = float(a)
print(f'a = {a} {type(a)}')
```

```python
b = 5
print(f'b = {b} {type(b)}')
print('Converting int to float....')

b = float(b)
print(f'b = {b} {type(b)}')
```

```python
c = 9 + 0j
print(f'c = {c} {type(c)}')
print('Converting complex number to float....')

c = float(c.real)
print(f'c = {c} {type(c)}')
```

## complex() : 

```python
e = '1'
print(f'e = {e} {type(e)}')
print('Converting str to complex....')

e = complex(e)
print(f'e = {e} {type(e)}')
```

```python
f = 3
print(f'f = {f} {type(f)}')
print('Converting int to complex....')

f = complex(f)
print(f'f = {f} {type(f)}')
```

```python
g = 7.98
print(f'g = {g} {type(g)}')
print('Converting float to complex....')

g = complex(g)
print(f'g = {g} {type(g)}')
```
