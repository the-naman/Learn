```python
from importlib.machinery import SourceFileLoader
connect_oracle = SourceFileLoader(
    "connect_oracle", "000. connect_oracle.py").load_module()
connection, cursor = connect_oracle.get_connection()
```

```python
cursor.execute("""
DECLARE
    v_msg VARCHAR2(100) := 'Success! File helper architecture is working perfectly.';
BEGIN
    DBMS_OUTPUT.PUT_LINE(v_msg);
END;
""")

connect_oracle.fetch_output(cursor)
```
