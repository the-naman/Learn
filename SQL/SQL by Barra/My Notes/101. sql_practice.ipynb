<script>
    function updateProgress() {
        const checkboxes = document.querySelectorAll('.todo-container input[type="checkbox"]');
        let checkedCount = 0;
        let savedState = {};

        checkboxes.forEach(cb => {
            savedState[cb.id] = cb.checked;
            if (cb.checked) {
                checkedCount++;
            }
        });

        const total = checkboxes.length;
        const percentage = total > 0 ? Math.round((checkedCount / total) * 100) : 0;

        document.getElementById('progress-bar').style.width = percentage + '%';
        document.getElementById('progress-text').innerText = percentage + '% (' + checkedCount + '/' + total + ' Completed)';

        // Persist records into system LocalStorage under a unique namespace
        localStorage.setItem('sql_roadmap_c4_tracker', JSON.stringify(savedState));
    }

    function loadSavedProgress() {
        const savedData = localStorage.getItem('sql_roadmap_c4_tracker');
        if (savedData) {
            const savedState = JSON.parse(savedData);
            for (const id in savedState) {
                const cb = document.getElementById(id);
                if (cb) {
                    cb.checked = savedState[id];
                }
            }
        }
        updateProgress();
    }

    function clearAllProgress() {
        if (confirm("Are you sure you want to reset your Category 4 progress?")) {
            const checkboxes = document.querySelectorAll('.todo-container input[type="checkbox"]');
            checkboxes.forEach(cb => cb.checked = false);
            updateProgress();
        }
    }

    // Run safe background loader for notebook engine layout instances
    setTimeout(loadSavedProgress, 150);
</script>
